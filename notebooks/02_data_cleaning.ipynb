```python
def treat_price(x):
    if type(x) == float:
        return x
    else:
        if x[1] == 'Lakh':
            return round(float(x[0])/100,2)
        else:
            return round(float(x[0]),2)
```

```python
def treat_area(y):
    if type(y) == float:
        return y
    
    num_str = y[0].replace(',', '')
    if y[1] == 'Marla':
        return round(float(num_str), 2)
    else:  # Kanal
        return round(float(num_str) * 20, 2)
        
```

```python
# Check how many duplicate rows exist across core listing details
duplicates_count = df_residential.duplicated(subset=['title', 'price', 'area_marla', 'sector', 'city']).sum()
print(f"Duplicate rows detected: {duplicates_count} out of {len(df_residential)}")
```

```python
# 1. Drop duplicates based on listing URL directly on df_residential
df_residential = (
    df_residential.drop_duplicates(subset=['link'])
    .copy()
    .reset_index(drop=True)
)

# 2. Filter out non-residential keywords and cap area/price bounds
residential_mask = (
    ~df_residential['title'].str.contains(
        r'commercial|plaza|shop|office|building|agri|farm',
        case=False,
        na=False,
    )
    & (df_residential['area_marla'] <= 40)
    & (df_residential['price'] <= 40)
)

df_residential = df_residential[residential_mask].copy().reset_index(drop=True)

# 3. Verify clean dataset shape and unique sectors
print(f"Final clean residential listings: {len(df_residential)}")
print(f"Unique sectors to geocode: {df_residential['sector'].nunique()}")
```

```python

# 1. Deduplicate by unique listing URL directly on df_residential
df_residential = (
    df_residential.drop_duplicates(subset=['link'])
    .copy()
    .reset_index(drop=True)
)

print(f"Clean unique residential listings: {len(df_residential)}")
print(f"Unique sectors to geocode: {df_residential['sector'].nunique()}")

# 2. Extract unique sector + city pairs for rate-limited geocoding
geo_df = (
    df_residential[['sector', 'city']]
    .drop_duplicates()
    .dropna()
    .reset_index(drop=True)
)
geo_df['search_query'] = (
    geo_df['sector'] + ', ' + geo_df['city'] + ', Pakistan'
)

# 3. Setup Nominatim geocoder
from geopy.extra.rate_limiter import RateLimiter
from geopy.geocoders import Nominatim

geolocator = Nominatim(user_agent="twin_cities_real_estate_enrichment")
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1.1)

coords = {}
print(f"Starting geocoding for {len(geo_df)} unique sectors...")

for idx, query in enumerate(geo_df['search_query'], start=1):
  try:
    loc = geocode(query)
    if loc:
      coords[query] = (loc.latitude, loc.longitude)
    else:
      # Fallback: strip sub-sectors (e.g. 'E-12/1' -> 'E-12')
      fallback_sector = query.split(',')[0].split('/')[0].strip()
      city_name = query.split(',')[1].strip()
      loc = geocode(f'{fallback_sector}, {city_name}, Pakistan')
      if loc:
        coords[query] = (loc.latitude, loc.longitude)
      else:
        coords[query] = (None, None)
  except Exception:
    coords[query] = (None, None)

  if idx % 25 == 0:
    print(f"Processed {idx}/{len(geo_df)} sectors...")

# 4. Map coordinates back onto df_residential
lookup = (
    df_residential['sector'].astype(str)
    + ', '
    + df_residential['city'].astype(str)
    + ', Pakistan'
)
df_residential['latitude'] = lookup.map(
    lambda x: coords.get(x, (None, None))[0]
)
df_residential['longitude'] = lookup.map(
    lambda x: coords.get(x, (None, None))[1]
)

print('\nGeocoding complete!')
print(
    f"Rows with coordinates: {df_residential['latitude'].notnull().sum()} /"
    f' {len(df_residential)}'
)
df_residential.head()
```

```python
remaining_nan_sectors = (
    df_residential[df_residential['latitude'].isnull()]['sector']
    .value_counts()
)
print(f"Remaining distinct sectors needing coordinates: {len(remaining_nan_sectors)}")
print("\nBreakdown of remaining sectors:")
print(remaining_nan_sectors.head(25))
```

```python
# 1. Drop any remaining NaNs in coordinates
df_valid = df_residential.dropna(subset=['latitude', 'longitude']).copy()

# 2. Compute price per marla for each listing
df_valid['price_per_marla'] = round(df_valid['price'] / df_valid['area_marla'], 4)

# 3. Aggregate by sector
sector_summary = df_valid.groupby(['city', 'sector']).agg(
    avg_price=('price', 'mean'),
    median_price=('price', 'median'),
    avg_price_per_marla=('price_per_marla', 'mean'),
    listing_count=('price', 'count'),
    latitude=('latitude', 'first'),
    longitude=('longitude', 'first')
).reset_index()

# Round financial metrics for clean tooltips
sector_summary['avg_price'] = sector_summary['avg_price'].round(2)
sector_summary['median_price'] = sector_summary['median_price'].round(2)
sector_summary['avg_price_per_marla'] = sector_summary['avg_price_per_marla'].round(3)

# 4. Save to CSV for the Streamlit app
sector_summary.to_csv('../data/processed/sector_price_summary.csv', index=False)
print(f"Exported summary for {len(sector_summary)} sectors successfully!")
```
